5. Apresente e depois elimine os outliers do campo salário. Considere que um outlier será dado pelo valor da média +- 2 vezes o seu desvio padrão.

```python

df.describe(include='all')
```

```python
limite_inferior = df['Salary'].mean() - 2 * df['Salary'].std()
limite_superior = df['Salary'].mean() + 2 * df['Salary'].std()

print('limite inferior:', limite_inferior)
print('limite superior:', limite_superior)
print('Dados fora dos limites:')
df[(df['Salary'] < limite_inferior) | (df['Salary'] > limite_superior)]
```

```python
df = df[(df['Salary']<=limite_superior) & (df['Salary']>=limite_inferior)]

df.describe(include='all')
```

6.Verifique se há alguma dado inconsistente considerando os campos idade e anos de experiência. Considere que as pessoas só podem começar a trabalhar após os 16 anos de idade.

```python
df['Inconsistent'] = ((df['Age'] - df['Years of Experience']) < 16)
df[df['Inconsistent'] == True]
```

```python
df.drop(labels=df[df['Inconsistent'] == True].index, inplace=True)
```
